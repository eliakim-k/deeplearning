### Know a bit about OpenMP
The OPM (OpenMP) is an API (Application Programming Interface) for shared memory multiprocessing programming in C, C++, and Fortran. It allows developers to write parallel programs that can run on multicore CPUs and shared memory systems, without requiring low-level threading primitives.

OpenMP provides a set of compiler directives, runtime library routines, and environment variables that enable the developer to specify which parts of the code should be executed in parallel, and how many threads should be used. The API is widely used in scientific computing, numerical simulation, and other applications that require high performance on shared memory systems.

OpenMP is designed to be portable across different operating systems and hardware architectures, and is supported by most major compilers, including GCC, Clang, Intel C++ Compiler, and Microsoft Visual Studio. The latest version of OpenMP, version 5.1, was released in November 2020, and includes new features for tasking, memory management, and GPU acceleration.

## Déclarer l'architecture du réseau de neurones

```python
modele = Sequential()
modele.add(Dense(2, input_dim=1, activation='relu'))
modele.add(Dense(1, activation='relu'))
modele.summary() # a keras method that gives a summaty of the network
```

### Poids de la couche 0 : definir_poids(model, couche, rang, coeff list, biais)

```python
definir_poids(modele, 0, 0, [1,], -1) # premier neurone
definir_poids(modele, 0, 1, [-0.5,], 1) # deuxième neurone
affiche_poids(modele, 0)
```

### Poids de la couche 1 : definir_poids(model, couche, rang, coeff, biais)

```python
definir_poids(modele, 1, 0, [1,1], 0)
affiche_poids(modele, 1)
```
