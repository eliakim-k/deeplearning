### Mutiplication and division methods

Here are some methods for multiplication and division of tensors in PyTorch:

- `torch.matmul()`: Performs matrix multiplication on two tensors.
- `torch.mul()`: Performs element-wise multiplication on two tensors.
- `torch.Tensor.multiply()`: Multiplies a tensor by a scalar value.
- `torch.div()`: Performs element-wise division on two tensors.

Here's an example code snippet that shows how to use these methods:


```
This will output:
```
```
tensor([[19, 22],
        [43, 50]])
tensor([[ 5, 12],
        [21, 32]])
tensor([[2, 4],
        [6, 8]])
tensor([[0.2000, 0.3333],
        [0.4286, 0.5000]])
```

````python
tensor1 = torch.tensor([[1, 2], [3, 4]])
tensor2 = torch.tensor([[5, 6], [7, 8]])

# Matrix multiplication
result1 = torch.matmul(tensor1, tensor2)

# Element-wise multiplication
result2 = torch.mul(tensor1, tensor2)

# Multiply a tensor by a scalar value
result3 = tensor1.multiply(2)

# Element-wise division
result4 = torch.div(tensor1, tensor2)

print(result1)
print(result2)
print(result3)
print(result4)
````

#### Create a numpy.ndarray from a tensor : .numpy() method

````python
# the method .numpy() converts a tensor into a numpy.ndarray class data
# the object still points to the same memory object : if one of the instance changes, the other also does.
arr = t.numpy()
print(arr,'\n', type(t),'\n', type(arr))
````

#### Create a tensor from numpy.ndarray : .from_numpy() method

````python
a = np.ones((3,3))
a.dtype
print(a)
````

````python
b = torch.from_numpy(a)
b.dtype
print(b)
````

````python
# if we modify the array, the same applies to the tensor automatically
a += 1 # here we modify the array
b # here the tensor has undergone the same changes
````
