1) Design model (input, output, forward pass with different layers)
2) Construct loss and optimizer
3) Training loop
- Forward = compute prediction and loss
- Backward = compute gradients
- Update weights

```python
import torch
import torch.nn as nn
```

## 0) Training samples

```python
# Linear regression
# f = w * x 

# here : f = 2 * x

# 0) Training samples
X = torch.tensor([[1], [2], [3], [4]], dtype=torch.float32)
Y = torch.tensor([[2], [4], [6], [8]], dtype=torch.float32)
X.shape
X_test = torch.tensor([5], dtype=torch.float32)
```

## 1) Design Model: Weights to optimize and forward function

```python
# 1) Design Model: Weights to optimize and forward function
input_size, output_size = X.shape[1], X.shape[1]

model = nn.Linear(in_features=input_size, out_features=output_size, bias=True)

print(f'Prediction before training: f(5) = {model(X_test).item():.3f}')
w, b = model.parameters()
b
```

```python
# Custom model

class LinearRegression(nn.Module):
    def __init__(self, input_dim, output_dim):
        #super().__init__()
        # Define layers
        self.lin = nn.Linear(input_dim, output_dim)
    
    def forward(self, x):
        return self.lin(x)

modelbis = LinearRegression(input_size, output_size)
dir(modelbis)
```

## 2) Define loss and optimizer

```python
# 2) Define loss and optimizer
learning_rate = 0.01
n_iters = 100

# callable function
loss = nn.MSELoss()

#optimizer = torch.optim.SGD(model.parameters(), lr=learning_rate)
optimizer = torch.optim.SGD(modelbis.parameters(), lr=learning_rate)
```

## 3) Training loop

```python
# 3) Training loop
for epoch in range(n_iters):
    # predict = forward pass
    y_predicted = modelbis(X)

    # loss
    l = loss(Y, y_predicted)

    # calculate gradients = backward pass
    l.backward()

    # update weights
    optimizer.step()

    # zero the gradients after updating
    optimizer.zero_grad() # different from "w.grad.zero_()"

    if epoch % 10 == 0:
        w, b = modelbis.parameters()
        print('epoch ', epoch+1, ': w = ', w[0,0].item(), ' loss = ', l.item())

print(f'Prediction after training: f(5) = {modelbis(X_test).item():.3f}')
```
