# 11. PyTorch Tutorial 11 - Softmax and Cross Entropy

```python
import numpy as np
X = np.linspace(0, 10, 9)
X = np.reshape(X, (3,3))
X
```

## Softmax function with arrays

```python
# # pay attention to the kwarg 'axis' : sum of exponentials over the column vector 
Y = np.exp(X) / np.sum(np.exp(X), axis=0)
Y
```

```python
Y[0,0] == np.exp(0.)/np.sum(np.exp(X[:,0])), Y[0,1] == np.exp(1.25)/np.sum(np.exp(X[:,1])), 
```

## What if I do not precise the axis? The sum takes into account all the elements of the array

```python
Ybis = np.exp(X) / np.sum(np.exp(X))
Ybis
```

```python
Y[0,0] == np.exp(0.)/ np.sum(np.exp(X))
```

## Softmax with torch

```python
import torch
```

```python
# pay attention to the kwarg 'dim'
T = torch.from_numpy(X)
torch.softmax(T, dim=0)
```

```python
T[0,0] == X[0, 0]
```
