# 5.2. Gradient Descent with gradient computed by torch

```python
import numpy as np
import torch
```

## Function : y = w * x

```python
# f = w * x

# f = 2 * x

X = torch.tensor([i+1 for i in range(15)], dtype=torch.float32)
Y = torch.tensor([i*2+2 for i in range(15)], dtype=torch.float32)
Y, X
```

### Intialize the model parameter

```python
w = torch.tensor(0.0, requires_grad=True)
w
```

## Model prediction

### Define the forward function, the loss and the gradient

```python
# model prediction
def forward(x):
    return w * x 

# loss : MSE
def loss(y, y_predicted):
    return ((y_predicted -y)**2).mean()

# gradients = backward
    # MSE : 1/N (w*x - y)**2
    # dJ/dw : 1/N 2*x (w*y - y)
    #def gradient(x, y, y_predicted):
        #return np.dot(2*x, y_predicted-y).mean()

print(f"Prediction before training f(5) = {forward(5)}")
```

### Train the model : the model fails when the X, Y vectors are larger than 15 elements

```python
# Training
learning_rate = 0.01
n_iters = 20

for epoch in range(n_iters):
    # prediction = forward pass
    y_pred = forward(X)

    # loss
    l = loss(Y, y_pred)

    # gradient = backward()
        # dw = gradient(X, Y, y_pred)
    l.backward()

    # update weights
    # Avoid to track(keep history of gradients during gradient descent)
    # this " with torch.no_grad()" is called a context manager
    with torch.no_grad():
        w -= learning_rate * w.grad
    # empty the w.grad after each iteration
    w.grad.zero_()

    if epoch % 2 == 0:
        print(f" epoch {epoch + 1} : w = {w:.3f}, loss = {l:.12f}")

print(f"Prediction after training f(5) = {forward(5)}")
```
