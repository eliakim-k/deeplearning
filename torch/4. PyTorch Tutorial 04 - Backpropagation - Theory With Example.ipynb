# 4. PyTorch Tutorial 04 - Backpropagation - Theory With Example

```python
import torch
```

 
 
 We are trying to compute the gradient of the above loss function with respect to w.

## Initial values for our variables

```python
x = torch.tensor(1.0)
y = torch.tensor(2.0)
w = torch.tensor(1.0, requires_grad=True) # We need to specify the kwarg "requires_grad=True"
```

## Forward pass and compute the loss

```python
y_hat = w * x
```

```python
loss = (y_hat - y)**2
loss
```

## Backward pass and retreive the gradient

```python
loss.backward() # Torch computes the local gradients and the final result using the chain rule
```

```python
w.grad # grad is an attribute of the variable that "required" gradient
```
