# Python : numpy et matplotlib avec une variable

```python
import numpy as np
import matplotlib.pyplot as plt
```

## La fonction cosinus

```python
X = np.linspace(0,10, num=81)
Y = np.cos(X)
```

```python
plt.plot(X, Y)
plt.show()
```

## Fonction amortie

```python
def amorti(x):
    return np.exp(-x) * np.cos(2 * np.pi * x)
Xprime = np.linspace(0,10,num=1000)
plt.plot(Xprime, amorti(Xprime))
plt.grid()
plt.show()
```

## Courbe de Gauss

```python
def gauss(x):
    return np.exp(-x**2)
```

```python
I = np.linspace(-3,3, num=1000)
F = gauss(I)
plt.plot(I, F)
plt.show()
```

## .plot() vs .scatter()

```python
R = np.random.randn(10)
O = np.array([np.random.randint(10) for i in range(10)])
O
```

```python
plt.scatter(O, R)
```

```python
plt.plot(O, R)
```

```python
plt.plot(O, R, linewidth=3, color='red')
```

## Axes

```python
plt.scatter(O, R)
plt.axis('equal')  # répère orthornormé
plt.xlim(2,5) # bornes de l'intervalle de l'axe des x qui sera visible sur le graphique
plt.title('Mon beau graphique') # titre du graphique
plt.grid() # ajouter une grille dans mon graphique
plt.show()
```

## Plusieurs tracées

```python
# Defining the function
def sigmoid(x, k):
    return 1/(1+np.exp(-x*k))
# Vector
V = np.linspace(-5, 5, num=100)
# Colors
colors = ['red', 'blue', 'green', 'pink']
# For-loop
for k in range(1, 5):
    W = sigmoid(V, k)
    plt.plot(V, W, color=colors[k-1], label=f"k={k}")
# the method .legend() displays the labels
plt.legend()
plt.show()
```
