# [Deepmath] 4.2. Matplotlib (deux variables)

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
```

```python
def f1(x, y):
    return x**2 - y**2

def f2(x, y):
    return 1/(2 + x**2 + y**2)

def f3(x, y):
    return x**2 + y**2

def f4(x, y):
    return np.absolute(x) + np.absolute(y)

def f5(x, y):
    r = np.sqrt(x**2 + y**2)
    z = np.sin(r)
    return z

def f6(x, y):
    return np.sin(x) + np.cos(y)
```

```python
n = 50
VX = np.linspace(-7, 7, n)
VY = np.linspace(-7, 7, n)
X, Y = np.meshgrid(VX, VY)
Z1 = f1(X, Y)
Z2 = f2(X, Y)
Z3 = f3(X, Y)
Z4 = f4(X, Y)
Z5 = f5(X, Y)
Z6 = f6(X, Y)
```

# Plotting the graphs : ax.plot_surface() ; ax.plot_wireframe()

## Point de selle $ f(x, y) = x^2 - y^2$

```python
fig1 = plt.figure()
ax1 = plt.axes(projection='3d')
ax1.view_init(40, -30)
ax1.plot_surface(X, Y, Z1)
# ax.plot_wireframe(X, Y, Z)
plt.show()
```

## Fonction $ f(x, y) = 1 / (4 + x^2 + y^2) $

```python
fig2 = plt.figure()
ax2 = plt.axes(projection='3d')
ax2.view_init(40, -70) # Orienation of the graph
ax2.plot_wireframe(X, Y, Z2)
plt.show()
```

## Fonction $ f(x, y) = x^2 + y^2 $

```python
fig3 = plt.figure()
ax3 = plt.axes(projection='3d')
ax3.view_init(40, -30) # Orienation of the graph
ax3.plot_wireframe(X, Y, Z3)
plt.show()
```

## Different ways to color the graphs

### The kwarg of 'ax.plot_surface()' method : cmap='hot'

```python
fig4 = plt.figure()
ax4 = plt.axes(projection='3d')
ax4.view_init(40, -60) # Orienation of the graph
my_graph4 = ax4.plot_surface(X, Y, Z4, cmap='hot')
fig4.colorbar(my_graph4)
plt.show()
```

### The kwarg : cmap='viridis'

```python
fig5 = plt.figure()
ax5 = plt.axes(projection='3d')
# Orienation of the graph
ax5.view_init(60, -40) 
# the first angle (in degree) is the 'elevation' : up-down rotation ; 
# the second is the 'azimut' : left-right rotation)
my_graph5 = ax5.plot_surface(X, Y, Z5, cmap='viridis')
fig5.colorbar(my_graph5)
plt.show()
```

# Les lignes de niveaux : 'plt.contour()'

```python
fig6 = plt.figure()
ax6 = plt.axes(projection='3d')
ax6.view_init(60, -50)
my_graph6 = ax6.plot_surface(X, Y, Z6, cmap="viridis")
fig6.colorbar(my_graph6)
plt.show()
```

```python
fig6 = plt.figure()
nb_niveaux = 5
plt.contour(X, Y, Z6, nb_niveaux)
plt.axis('equal') # fixant le repère comme orthonormé
plt.show()
```
