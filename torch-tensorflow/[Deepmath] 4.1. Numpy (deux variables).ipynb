```python
# Browse all the elements of the matrix column-wise
for j in range(A.shape[1]):
    for i in range(A.shape[0]):
        print(A[i,j])
```

```python
# Browse all the elements of the matrix row-wise
for i in range(A.shape[0]):
    for j in range(A.shape[1]):
        print(A[i,j])
```

# Cartesian product to make a 2D-grid

```python
VX = np.linspace(0,2, num=5)
VY = np.linspace(0,2, num=5)
# Drawing a grid with a list comprehension and a for-loop
CP = np.array([(x, y) for x in VX for y in VY])
print(CP)
for point in CP:
    plt.scatter(point[0], point[1])
plt.show()
```

# Making a similar grid with one numpy method : np.meshgrid(vector1, vector2)

```python
X, Y = np.meshgrid(VX, VY)
print(f'X = \n {X}', "\n",f'Y = \n {Y}')
```

```python
plt.scatter(X, Y)
plt.show()
```

# Evaluate a function for each point on the grid

```python
def f(x, y):
    return x**2 + y**2

Z = f(X, Y)
Z
```
