# Tensors of Tensorflow : Calcul du gradient

```python
import tensorflow as tf
import numpy as np
```

## Définir une fonction

```python
def f(x, y):
    return x**2 - y**2
```

## Calcul du gradient en (x0, y0) = (2, 3)

```python
x, y = tf.Variable(2.), tf.Variable(3.)
x
```

```python
with tf.GradientTape() as tape:
    z = f(x, y)
```

```python
tfgradient = tape.gradient(z, [x, y])
gradient = [tfgradient[i].numpy() for i in range(len(tfgradient))]
gradient
```

## Calcul du gradient en plusieurs points

```python
x = tf.range(1, 4, 0.5)
```

```python
def g(x):
    return 2 * tf.math.log(x) + tf.math.exp(-x)
```

```python
with tf.GradientTape() as tape:
    tape.watch(x)
    y = g(x)
```

```python
tfgradient = tape.gradient(y, x)
gradient = [tfgradient[i].numpy() for i in range(len(tfgradient))]
gradient
```
