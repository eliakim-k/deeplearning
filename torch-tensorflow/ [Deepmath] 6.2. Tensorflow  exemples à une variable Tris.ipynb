## Pouver le théorème de l'approximation universelle

Théorème 2 (Théorème d’approximation universelle).
Toute fonction continue f : [a, b] → R peut être approchée d’aussi près que l’on veut par une fonction
F : [a, b]→R réalisée par un réseau de neurones.

### La fonction à approcher

```python
def f(x):
    return x * np.sin(x) / (1 + x**2)
a = -10
b = 10
n = 100
VX = np.linspace(a, b, n)
VY = f(VX)
plt.plot(VX, VY)
plt.show()
```

### Architecture du réseau

```python
model = Sequential()
model.add(Dense(2*n, input_dim=1, activation=heaviside))
model.add(Dense(1, activation='linear'))
model.summary()
```

```python
calcul_approximation(model, f, a, b, n)
affichage_approximation(model, f, a, b)
```
